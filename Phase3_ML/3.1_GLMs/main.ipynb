## Build a Poisson distribution predictor

Dự đoán số lần khách hàng ghé thăm cửa hàng thông qua các chỉ số còn lại

```python
import pandas as pd
df = pd.read_csv("..\\dataset\\poisson_visits.csv")
df.head()
```

```python
# Data preprocessing
df['temperature'] = df['temperature'] - 30
df.head()
```

```python
# Modeling

import numpy as np

X = df.drop(columns = ["visits"]).to_numpy()
X = np.column_stack((np.ones(X.shape[0]), X)) # (n, d)

y = df[["visits"]].to_numpy() # (n, 1)

print(y.shape, X.shape)
```

```python
# Mini-batch gradient descent
learning_rate = 0.001
theta = np.zeros(X.shape[1]) 
theta = theta.reshape(-1, 1) # (d, 1)
epoch = 10000
mini_batch = 1000 # GD

for _ in range(epoch):
    start = 0
    while start <= X.shape[0] - 1:
        end = min(X.shape[0], start + mini_batch)
        x_now = X[start:end, :]
        y_now = y[start:end]

        y_pred = np.exp(x_now @ theta).reshape(-1, 1)

        gradient = (x_now.T @ (y_now - y_pred)).reshape(-1, 1)

        theta += learning_rate * gradient / (end - start)

        start = end

print(theta)
```
