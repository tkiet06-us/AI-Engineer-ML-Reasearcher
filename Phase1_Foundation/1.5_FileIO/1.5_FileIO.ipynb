# 2. Open and Read File

```python
# Context manager: tự động đóng file
with open("subject.txt", "r") as file:
    content = file.read()

print(content)
```

```python
with open("subject.txt", "r") as file:
    for line in file:
        print(line)
```

```python
with open("subject.txt", "r") as file:
    for line in file:
        print(line.strip())
```

# 3. Read CSV File

```python
import csv

with open("student.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    print(reader)

    for row in reader:
        print(row)
```

```python
with open("student.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    print(reader)

    for dict in reader:
        print(dict)
```

```python
fruits = [
    ["carrot", 3, 1000],
    ["apple", 6, 500],
    ["banana", 10, 300]
]
print(sorted(fruits, key=lambda fruit: fruit[0]))
```

```python
with open("fruits.txt", "w", encoding = "utf-8", newline = "") as file:
    writer = csv.writer(file)

    writer.writerow(["name", "quantity", "price"])

    writer.writerows(fruits)
# Nhớ thêm newline="" để tránh trường hợp writer tự thêm 1 dòng trống sau mỗi dòng được ghi
```

```python
fruits = [
    {"name": "apple", "quantity": 6, "price": 10},
    {"name": "banana", "quantity": 16, "price": 14}
]

with open("fruits.txt", "w", encoding = "utf-8", newline = "") as file:
    fieldsname = ["name", "quantity", "price"]
    writer = csv.DictWriter(
        file,
        fieldnames=fieldsname
    )

    writer.writeheader()
    writer.writerows(fruits)
```

# 4. JSON

```python
import json

with open("config.json", encoding="utf-8") as file:
    config = json.load(file)

print(config)
```

```python
import json

config = {
    "model": "gpt-oss-20b",
    "batch_size": 100,
    "learning_rate": 0.002
}

with open("config.json", "w", encoding="utf-8") as file:
    json.dump(config, file)
```

```python
import json

with open("config.json", encoding="utf-8") as file:
    config = json.loads("{\"model\": \"gpt-oss-20b\", \"batch_size\": 100, \"learning_rate\": 0.002}")

print(config)

# Chuỗi JSON thành đối tượng Python
```

```python
import json

with open("config.json", "w", encoding="utf-8") as file:
    content = json.dumps(config, indent = 3)

print(content)

# Đối tượng Python thành chuỗi JSON
```

# 5. Pickle file

```python
students = {
  'Student 1': {
        'Name': "Alice", 'Age' :10, 'Grade':4,
    },   
    'Student 2': {
        'Name':'Bob', 'Age':11, 'Grade':5
    },   
    'Student 3': {
        'Name':'Elena', 'Age':14, 'Grade':8
    }   
}
```

```python
import pickle
with open("students.pkl", "wb") as file:
    pickle.dump(students, file)
```

```python
with open("students.pkl", "rb") as file:
    content = pickle.load(file)
print(content)
```
