## 1.6. Overriding methods

```python
import random
class Tool:
    total_tools_created = 0
    registry = []

    def __init__(self, name, description):
        self.name = name
        self.description = description
        self.create_at = datetime.now()
        Tool.total_tools_created += 1
        Tool.registry.append(self)

        print("[Tool created] name=" + str(self.name) + " at " + str(self.create_at))

    def run(self, input_data):
        print(f"Tool {self.name} received: {input_data}")
        return None

    def describe(self):
        print(f"[{self.name}] Thực hiện phép tính")

class CalculatorTool(Tool):
    def __init__(self, name, description) -> None:
        super().__init__(name, description)
        self.allowed_operators = ["+", "-", "*", "/"]

    def run(self, input_data):
        parts = input_data.split()
        if len(parts) != 3 or parts[1] not in self.allowed_operators:
            return "Invalid expression"
        a, op, b = float(parts[0]), parts[1], float(parts[2])
        result = 0
        if op == '+': result = a + b
        if op == '-': result = a - b
        if op == '*': result = a * b
        if op == '/': result = a / b if b != 0 else float('nan')
        return str(result)

class TranslatorTool(Tool):
    def __init__(self, name, description) -> None:
        super().__init__(name, description)
        self.dictionary = {'hello': 'xin chào', 'book': 'quyển sách'}

    def run(self, input_data):
        if input_data not in self.dictionary:
            return "Không tìm thấy bản dịch"
        return self.dictionary[input_data]

class WeatherTool(Tool):
    def __init__(self, name, description) -> None:
        super().__init__(name, description)
        self.unit = 'celsius'

    def run(self, input_data):
        print(f"Nhiệt độ tại {input_data}: {random.randint(15, 38)}*C")
        
```

## 1.7. @property, @staticmethod, @classmethod

```python
import random
class Tool:
    total_tools_created = 0
    registry = []

    def __init__(self, name, description):
        self.name = name
        self.description = description
        self.create_at = datetime.now()
        self._usage_count = 0
        Tool.total_tools_created += 1
        Tool.registry.append(self)

        print("[Tool created] name=" + str(self.name) + " at " + str(self.create_at))

    def run(self, input_data):
        print(f"Tool {self.name} received: {input_data}")
        return None

    def describe(self):
        print(f"[{self.name}] Thực hiện phép tính")

    @property
    def usage_count(self):
        return self._usage_count

    @staticmethod
    def validate_name(name):
        for char in name:
            if (char > 'z' or char < 'a') and (char < '0' or char > '9') and char != '_' and char != ' ':
                return False
        return True

    @classmethod
    def from_config(cls, config: dict):
        if cls.validate_name(config['name']):
            return cls(config['name'], config['description'])
        print("Tên không hợp lệ")
        return None
    
class CalculatorTool(Tool):
    def __init__(self, name, description) -> None:
        super().__init__(name, description)
        self.allowed_operators = ["+", "-", "*", "/"]

    def run(self, input_data):
        parts = input_data.split()
        if len(parts) != 3 or parts[1] not in self.allowed_operators:
            return "Invalid expression"
        a, op, b = float(parts[0]), parts[1], float(parts[2])
        result = 0
        if op == '+': result = a + b
        if op == '-': result = a - b
        if op == '*': result = a * b
        if op == '/': result = a / b if b != 0 else float('nan')
        self._usage_count += 1
        return str(result)

class TranslatorTool(Tool):
    def __init__(self, name, description) -> None:
        super().__init__(name, description)
        self.dictionary = {'hello': 'xin chào', 'book': 'quyển sách'}

    def run(self, input_data):
        if input_data not in self.dictionary:
            return "Không tìm thấy bản dịch"
        self._usage_count += 1
        return self.dictionary[input_data]

class WeatherTool(Tool):
    def __init__(self, name, description) -> None:
        super().__init__(name, description)
        self.unit = 'celsius'

    def run(self, input_data):
        print(f"Nhiệt độ tại {input_data}: {random.randint(15, 38)}*C")
        self._usage_count += 1
        
```

## 1.8. Abstract method

```python
import random
from abc import ABC, abstractmethod
class Tool(ABC):
    total_tools_created = 0
    registry = []

    @abstractmethod
    def __init__(self, name, description):
        self.name = name
        self.description = description
        self.create_at = datetime.now()
        self._usage_count = 0
        Tool.total_tools_created += 1
        Tool.registry.append(self)

        print("[Tool created] name=" + str(self.name) + " at " + str(self.create_at))

    def run(self, input_data):
        print(f"Tool {self.name} received: {input_data}")
        return None

    def describe(self):
        print(f"[{self.name}] Thực hiện phép tính")

    @property
    @abstractmethod
    def input_type(self):
        pass

    @property
    def usage_count(self):
        return self._usage_count

    @staticmethod
    def validate_name(name):
        for char in name:
            if (char > 'z' or char < 'a') and (char < '0' or char > '9') and char != '_' and char != ' ':
                return False
        return True

    @classmethod
    def from_config(cls, config: dict):
        if cls.validate_name(config['name']):
            return cls(config['name'], config['description'])
        print("Tên không hợp lệ")
        return None
    
class CalculatorTool(Tool):
    def __init__(self, name, description) -> None:
        super().__init__(name, description)
        self.allowed_operators = ["+", "-", "*", "/"]

    def run(self, input_data):
        parts = input_data.split()
        if len(parts) != 3 or parts[1] not in self.allowed_operators:
            return "Invalid expression"
        a, op, b = float(parts[0]), parts[1], float(parts[2])
        result = 0
        if op == '+': result = a + b
        if op == '-': result = a - b
        if op == '*': result = a * b
        if op == '/': result = a / b if b != 0 else float('nan')
        self._usage_count += 1
        return str(result)

    @property
    def input_type(self):
        return "math_expression"

class TranslatorTool(Tool):
    def __init__(self, name, description) -> None:
        super().__init__(name, description)
        self.dictionary = {'hello': 'xin chào', 'book': 'quyển sách'}

    def run(self, input_data):
        if input_data not in self.dictionary:
            return "Không tìm thấy bản dịch"
        self._usage_count += 1
        return self.dictionary[input_data]

    @property
    def input_type(self):
        return "text"

class WeatherTool(Tool):
    def __init__(self, name, description) -> None:
        super().__init__(name, description)
        self.unit = 'celsius'

    def run(self, input_data):
        print(f"Nhiệt độ tại {input_data}: {random.randint(15, 38)}*C")
        self._usage_count += 1

    @property
    def input_type(self):
        return "city_name"
        
```
