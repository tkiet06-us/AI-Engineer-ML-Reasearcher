```python
s = "I,love,HCMUS"
s.split(',')
```

```python
words = ['I', 'love', 'HCMUS']
s = " ".join(words)
s
```

```python
word = "       I        love       HCMUS      "
print(word.strip())
print(word.lstrip())
print(word.rstrip())
```

```python
word = "I love Python"
print(word.replace("love", "hate"))
```

```python
word = "I love you more than you love me"
word.replace("love", "hate", 1)
```

```python
word = "I love you more than you love me"
parts = word.rsplit("love", 1)
word = "hate".join(parts)
word
```
