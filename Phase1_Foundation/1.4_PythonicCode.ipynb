# 1. Comprehension

## 1.1 List

```python
lst = [x ** 2 for x in range(5)]
print(lst)
```

```python
import pprint
matrix = [
    [(i + j) % 5 for i in range(5)]
    for j in range(5)
]
pprint.pprint(matrix)
```

```python
even = [i for i in range(10) if i % 2 == 0]
print(even)
```

## 1.2. Dictionary

```python
squared = {x: x ** 2 for x in range(5)}
pprint.pprint(squared)
```

```python
double_even = {x: x * 2 for x in range(10) if x % 2 == 0}
print(double_even)
```

## 1.3. Set

```python
double = {x * 2 for x in range(10)}
print(double)
```

```python
rand_0_4 = {x % 5 for x in range(20)}
print(rand_0_4)
```

# 2. Generator

## 2.1. Expression

Generator được sử dụng khi thao tác với dữ liệu lớn, nó chỉ tạo các phần tử khi cần

```python
my_list = (x ** 2 for x in [1, 4, 8, 10])
print(my_list)
```

```python
for num in my_list:
    print(num)
```

```python
print(my_list[2])
```

```python
print(sum(my_list))
```

```python
sum(x**2 for x in range(10))
```

## 2.2. Function

```python
def generate():
    n = 0
    while True:
        yield n
        n += 1
```

```python
print(generate())
```

```python
a = generate()
print(next(a), next(a), next(a))
```

```python
def my_gen():
    n = 1
    print('Doan text nay duoc in dau tien')
    yield n

    n += 1
    print('Doan text nay duoc in thu hai')
    yield n

    n += 1
    print('Doan text nay duoc in cuoi cung')
    yield n
```

```python
for num in my_gen():
    print(num)
```

# 3. Map, filter, reduce

```python
numbers = [0, 1, 2, 3]
double = map(lambda x: x ** 2, numbers)
print(list(double))
```

```python
numbers = [0, 1, 2, 3, 4, 5]
even = map(lambda x: x % 2 == 0, numbers)
print(list(even))
```

```python
numbers = [0, 1, 2, 3, 4, 5]
even = filter(lambda x: x % 2 == 0, numbers)
print(list(even))
```

```python
from functools import reduce
numbers = [1, 2, 3, 4]
multiply = reduce(lambda x, y: x * y, numbers)
print(multiply)
```

Giải thích:
- $1 \times 2 = 2$
- $2 \times 3 = 6$
- $6 \times 4 = 24$

```python
words = ["i ", "love ", "hcmus"]
sentence = reduce(lambda x, y: x + y, words)
print(sentence)
```

```python
mul3 = reduce(lambda x, y, z: x + y + z, numbers)
print(mul3)
```

# 4. Sorted with key

Key là 1 phần tử trong list

```python
students = [
    ("An", 7),
    ("Cường", 9),
    ("Bình", 8)
]

result = sorted(students, key = lambda student: student[1])
print(result)
```

```python
result = sorted(students, key = lambda student: student[1], reverse = True)
print(result)
```

# 5. Collections

```python
from collections import Counter
letters = ['a', 'a', 'b', 'b', 'b', 'c']
counter = Counter(letters)
print(counter)
```

```python
counter.most_common(1)
```

```python
from collections import defaultdict

data = defaultdict(list)

data["python"].append('list')
data["python"].append('dict')

print(data)
```

```python
students = [
    ("A", "AI"),
    ("B", "Web"),
    ("C", "AI"),
    ("D", "Web")
]

groups = defaultdict(list)

for name, major in students:
    groups[major].append(name)

pprint.pprint(groups)
```

```python
from collections import deque
queue = deque()

for i in range(1, 6):
    queue.append(i)

print(queue)
```

```python
queue.append(6)
print(queue)
```

```python
queue.appendleft(0)
print(queue)
```

```python
queue.pop()
queue.popleft()
print(queue)
```
